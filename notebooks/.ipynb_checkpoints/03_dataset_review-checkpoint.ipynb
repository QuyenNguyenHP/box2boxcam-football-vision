Le test est proposé en premier parce qu'une métrique n'est crédible que si sa vérité terrain est propre.

## Signification des quatre décisions

| Bouton | Effet futur dans `v002` | Quand l'utiliser |
|---|---|---|
| **Garder le label** | conserve le label actuel | la boîte rouge encadre bien le ballon, ou le négatif est réellement sans ballon |
| **Passer en négatif** | écrit un label vide | la boîte rouge est fausse et aucun ballon n'est visible |
| **Sauver la boîte dessinée** | remplace le label par la boîte dessinée | ballon visible mais boîte absente, fausse ou imprécise |
| **Exclure de v002** | retire l'image | décision impossible : occultation, ambiguïté ou image inutilisable |

Les rectangles rouges sont les annotations d'origine. Pour redessiner, trace un rectangle à la souris sur l'image centrale, puis sauvegarde-le. Les images précédente et suivante donnent le contexte temporel mais ne sont pas modifiées.
